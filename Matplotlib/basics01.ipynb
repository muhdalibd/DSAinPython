<b style="color: #fa851e">Matplotlib</b><br>
Matplotlib is a Python library for creating static, animated and interactive visualizations.<br>
<b>Plot types</b><br>
<b style="color: #dd3290">1. Line plot</b><br>
A line plot connects data points with straight lines to show trends. In Matplotlib, the plot() function is used to create it.<br>
Common uses of line plots:<br>
&emsp; <span style="color: #343F50">Tracking time series data (e.g., stock prices, temperature).</span><br>
&emsp; <span style="color: #343F50">Comparing multiple datasets.</span><br>
&emsp; <span style="color: #343F50">Identifying patterns and trends in continuous data.</span><br>

<b style="color: #dd3290">2. Bar plot</b><br>
A bar plot uses rectangular bars to represent categorical data, with height indicating value. In Matplotlib, it's created using the bar() function.<br>
Common uses of bar plots:<br>
&emsp; <span style="color: #343F50">Comparing categories or groups.</span><br>
&emsp; <span style="color: #343F50">Showing distributions and frequencies.</span><br>
&emsp; <span style="color: #343F50">Visualizing differences in data values.</span><br>

<b style="color: #dd3290">3. Scatter plot</b><br>
A scatter plot visualizes relationships between two variables using individual data points. In Matplotlib, it's created with the scatter() function.<br>
Common uses of scatter plots:<br>
&emsp; <span style="color: #343F50">Showing relationships between two numerical variables.</span><br>
&emsp; <span style="color: #343F50">Identifying trends, clusters, and outliers.</span><br>
&emsp; <span style="color: #343F50">Visualizing correlations in data.</span><br>

<b style="color: #dd3290">4. Pie plot</b><br>
A pie plot represents data as slices of a circle, showing proportions of a whole. In Matplotlib, it's created using the pie() function.<br>
Common uses of pie plots:<br>
&emsp; <span style="color: #343F50">Showing proportions or percentage distribution.</span><br>
&emsp; <span style="color: #343F50">Visualizing categorical data as parts of a whole.</span><br>
&emsp; <span style="color: #343F50">Comparing relative sizes of different categories.</span><br>

<b style="color: #dd3290">5. Histogram</b><br>
A histogram represents the distribution of numerical data using bars, where height indicates frequency. In Matplotlib, it's created using the hist() function.<br>
Common uses of histograms:<br>
&emsp; <span style="color: #343F50">Visualizing the distribution of continuous data.</span><br>
&emsp; <span style="color: #343F50">Identifying patterns, skewness, and outliers.</span><br>
&emsp; <span style="color: #343F50">Understanding frequency distribution in datasets.</span><br>

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
The <span style="color: #343F50">plt.subplots()</span> function from ⁠Matplotlib is used to simultaneously create a container figure and one or more individual axes (subplots). It is the foundation of Matplotlib's object-oriented plotting interface.<br>
&emsp; <span style="color: #0e9dc5">fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(width, height), sharex=False, sharey=False)</span>
</div>

```python
import matplotlib.pyplot as plt
# Create the figure and a single axes
fig, ax = plt.subplots()
# 1 row, 2 columns of subplots
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
# Access subplots using a single index
ax[0].plot()
ax[1].plot()
# 2 rows, 2 columns of subplots
fig, axs = plt.subplots(2, 2, figsize=(8, 8))
axs[0, 0].plot()
```

```python
# a figure with one Axes on the left, and two on the right:
fig, axs = plt.subplot_mosaic([
    ['left', 'right_top'],
    ['left', 'right_bottom']
])
```
