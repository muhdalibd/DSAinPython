<h4 style="color:#fa851e">Random Number Generation and Statistics</h4>
<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
NumPy’s random module provides a list of functions for generating random numbers, which are essential for simulations, cryptography and machine learning applications.
</div>

<h5 style="color:#fa851e">Generating Random Numbers</h5>
<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
The <span style="color: #0e9dc5">numpy.random.randint()</span> function is used to generate random integers within a specified range.
<pre style="font-family: inherit; font-weight: 400; font-size: 16px; color: gray;">
        Syntax
        <span style="color: #0e9dc5">numpy.random.randint(low, high=None, size=None, dtype=int)</span>
        Parameters:
        low : Lowest integer that can appear in the output.
        high (optional) : Upper limit (exclusive). If omitted, range becomes [0, low).
        size (optional) : Shape of the output array (e.g., 5, (2,3), (2,3,4)).
        dtype (optional) : Data type of the returned numbers. Default is integer.
</pre>
</div>

```python
import numpy as np
arr = np.random.randint(0, 5, size=5)
print(arr)
```

```python
import numpy as np
arr = np.random.randint(0, 10, size=(2, 3))
print(arr)
```

```python
import numpy as np
arr = np.random.randint(5, 10, size=(2, 2, 4))
print(arr)
```

<h4 style="color: #0e9dc5">Normal Distribution in NumPy</h4>
<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
The Normal (Gaussian) Distribution is a commonly used probability distribution that models natural data such as test scores, heights, sensor readings and measurement variations.
In NumPy, we generate values from a Normal Distribution using the <span style="color: #0e9dc5">numpy.random.normal()</span> method, which makes it simple to create realistic, statistically consistent data for analysis and simulations.
<pre style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
    Syntax
    <span style="color: #0e9dc5">numpy.random.normal(loc=0.0, scale=1.0, size=None)</span>
    Parameters:
    loc: Mean (center) of the distribution.
    scale: Standard deviation (spread).
    size: Shape of the output (single value, list, matrix, etc.)
</pre>
</div>

```python
import numpy as np
x = np.random.normal()
print(x)
```

```python
arr = np.random.normal(size=5)
print(arr)
```

```python
arr = np.random.normal(loc=10, scale=2, size=(2, 3))
print(arr)
```

<h5 style="color: #0e9dc5">Visualizing the Normal Distribution</h5>


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generate data
data = np.random.normal(loc=0, scale=1, size=1000)

# Plot histogram
plt.hist(data, bins=30, edgecolor='black', density=True)

# Plot theoretical PDF
x = np.linspace(-4, 4, 200)
pdf = norm.pdf(x, loc=0, scale=1)
plt.plot(x, pdf, label="Theoretical PDF")

plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.grid(True)
plt.legend()
plt.show()
```

<h4 style="color: #0e9dc5">Binomial Distribution in NumPy</h4>
<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
The Binomial Distribution models the number of successes in a fixed number of independent trials where each trial has only two outcomes: success or failure. In NumPy, we use the <span style="color: #0e9dc5">numpy.random.binomial()</span> method to generate values that follow this distribution. It is commonly used in coin flips, defect detection, surveys, and probability experiments.
<pre style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
    Syntax
    <span style="color: #0e9dc5">numpy.random.binomial(n, p, size=None)</span>
    Parameters:
    n: Number of trials
    p: Probability of success in each trial
    size: Shape of output array
</pre>
</div>

```python
import numpy as np
x = np.random.binomial(n=10, p=0.5)
print(x)
```

```python
import numpy as np
arr = np.random.binomial(n=10, p=0.5, size=5)
print(arr)
```

```python
import numpy as np
m = np.random.binomial(12, 0.6, size=(2, 3))
print(m)
```

<h5 style="color: #0e9dc5">Visualizing the Binomial Distribution</h5>

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.5
size = 1000

data = np.random.binomial(n, p, size)

plt.hist(data, bins=np.arange(-0.5, n+1.5, 1), density=True, edgecolor='black', alpha=0.7, label='Histogram')

x = np.arange(0, n+1)
pmf = binom.pmf(x, n, p)

plt.scatter(x, pmf, color='red', label='Theoretical PMF')
plt.vlines(x, 0, pmf, colors='red', linestyles='dashed')

plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.legend()
plt.grid(True)
plt.show()
```

<h4 style="color: #0e9dc5">Poisson Distribution in NumPy</h4>
<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
The Poisson Distribution models how many times an event occurs within a fixed interval when the average occurrence rate (λ) is known. It is commonly used for scenarios like customer arrivals, call center traffic, website visits, or any event that happens independently over time or space.
In NumPy, we use the <span style="color: #0e9dc5">numpy.random.poisson()</span> method to generate Poisson-distributed random values.
<pre style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
    Syntax
    <span style="color: #0e9dc5">numpy.random.poisson(lam=1.0, size=None)</span>
    Parameters:
    lam: Average expected events (λ).
    size: Shape of output (e.g., single value, 1D array, 2D array).
</pre>
</div>

```python
import numpy as np
x = np.random.poisson(lam=3)
print(x)
```

```python
arr = np.random.poisson(lam=5, size=5)
print(arr)
```

```python
m = np.random.poisson(lam=4, size=(2, 3))
print(m)
```

<h5 style="color: #0e9dc5">Visualizing the Poisson Distribution</h5>

```python
import numpy as np
import matplotlib.pyplot as plt

lam = 3
data = np.random.poisson(lam=lam, size=1000)

plt.hist(data, bins=np.arange(-0.5, max(data)+1.5, 1), edgecolor='black', density=True)
plt.title(f"Poisson Distribution (λ={lam})")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.grid(True)
plt.show()
```

<h4 style="color: #0e9dc5">Uniform Distribution in NumPy</h4>
<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
A Uniform Distribution is used when every value in a given range has an equal probability of occurring. NumPy provides the <span style="color: #0e9dc5">numpy.random.uniform()</span> method to generate such values for simulations, sampling, and numerical experiments.
<pre style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
    Syntax
    <span style="color: #0e9dc5">numpy.random.uniform(low=0.0, high=1.0, size=None)</span>
    Parameters:
    low: Lower bound of the range (inclusive).
    high: Upper bound of the range (exclusive).
    size: Shape of the output array.
</pre>
</div>

```python
import numpy as np
num = np.random.uniform()
print(num)
```

```python
arr = np.random.uniform(size=5)
print(arr)
```

```python
vals = np.random.uniform(10, 20, size=5)
print(vals)
```

```python
rng = np.random.default_rng()  # Create a Generator instance
m = rng.uniform(1, 5, size=(2, 3))
print(m)
```

<h5 style="color: #0e9dc5">Visualizing the Uniform Distribution</h5>

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

low = 10
high = 20
size = 1000

data = np.random.uniform(low, high, size)

sns.histplot(data, bins=30, kde=False, color='skyblue', edgecolor='black')

plt.title(f"Uniform Distribution (Range: {low} to {high})")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

<h4 style="color: #0e9dc5">Exponential Distribution in NumPy</h4>
<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
The Exponential Distribution is a continuous probability distribution that describes the time between two events in a Poisson process, where events occur independently and at a constant average rate. NumPy provides a simple method to generate such random values: <span style="color: #0e9dc5">numpy.random.exponential()</span>.
<pre style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
    Syntax
    <span style="color: #0e9dc5">numpy.random.exponential(scale=1.0, size=None)</span>
    Parameters:
    scale: Inverse of the event rate (β = 1/λ).
    size: Shape of output array.
</pre>
</div>

```python
import numpy as np
x = np.random.exponential()
print(x)
```

```python
x = np.random.exponential(scale=2)
print(x)
```

```python
arr = np.random.exponential(scale=1.5, size=5)
print(arr)
```

```python
arr = np.random.exponential(scale=1.5, size=(2, 3))
print(arr)
```

<h5 style="color: #0e9dc5">Visualizing the Exponential Distribution</h5>

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

s = 2      # scale
n = 800    # number of points

data = np.random.exponential(scale=s, size=n)
sns.histplot(data, bins=30, kde=True, edgecolor='black')

plt.title(f"Exponential Distribution (Scale={s})")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

<h4 style="color: #0e9dc5">Chi-Square Distribution in NumPy</h4>
<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
The Chi-Square Distribution appears when you add up the squares of independent standard normal random variables. It is widely used in hypothesis testing, goodness-of-fit tests, variance testing, and statistical modeling. In NumPy, we generate Chi-Square values using <span style="color: #0e9dc5">numpy.random.chisquare()</span>.
<pre style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
    Syntax
    <span style="color: #0e9dc5">numpy.random.chisquare(df, size=None)</span>
    Parameters:
    df: Degrees of freedom (controls the shape of the curve)
    size: Shape of output array
</pre>
</div>

```python
import numpy as np
x = np.random.chisquare(df=2)
print(x)
```

```python
arr = np.random.chisquare(df=2, size=5)
print(arr)
```

```python
x = np.random.chisquare(5, size=4)
print(x)
```

```python
m = np.random.chisquare(3, size=(2, 3))
print(m)
```

<h5 style="color: #0e9dc5">Visualizing the Chi-Square Distribution</h5>

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2

df = 2
size = 1000

data = np.random.chisquare(df, size)
plt.hist(data, bins=30, density=True, edgecolor='black', alpha=0.7, label='Histogram')

x = np.linspace(0, max(data), 200)
pdf = chi2.pdf(x, df)
plt.plot(x, pdf, color='red', label='Theoretical PDF')

plt.title(f"Chi-Square Distribution (df={df})")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.grid(True)
plt.show()
```

<h4 style="color: #0e9dc5">Statistical Functions in NumPy</h4>
<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
In NumPy, we can compute the mean, standard deviation, and variance of a given array along the second axis by two approaches first is by using inbuilt functions and second is by the formulas of the mean, standard deviation, and variance.
<pre style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
    Method 1: Using numpy.mean(), numpy.std(), numpy.var()
    Method 2: Using the formulas 
</pre>
</div>

```python
import numpy as np

# Original array
array = np.arange(10)
print(array)

r1 = np.mean(array)
print("Mean: ", r1)

r2 = np.std(array)
print("std: ", r2)

r3 = np.var(array)
print("variance: ", r3)
```

```python
import numpy as np

# Original array
array = np.arange(10)
print(array)

r1 = np.average(array)
print("Mean: ", r1)

r2 = np.sqrt(np.mean((array - np.mean(array)) ** 2))
print("std: ", r2)

r3 = np.mean((array - np.mean(array)) ** 2)
print("variance: ", r3)
```
