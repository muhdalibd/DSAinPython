<span style="color:#fa851e; font-size: 20px">Python Tuples</span><br>
A tuple is an immutable ordered collection of elements.<br>
&emsp; Tuples are similar to lists, but unlike lists, they cannot be changed after their creation.<br>
&emsp; Can hold elements of different data types.<br>
&emsp; These are ordered, heterogeneous and immutable.<br>
</duv>

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #0e9dc5">Creating a Tuple</span><br>
A tuple is created by placing all the items inside parentheses (), separated by commas. A tuple can have any number of items.<br>
1. Using Parentheses () (Standard Syntax)
The most common way to create a tuple.

```python
empty_tuple = ()
numbers = (1, 2, 3)
mixed_types = (1, "hello", 3.14, True)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">2. Single-Element Tuple (Trailing Comma Required)</span><br>
To create a tuple with only one element, you must include a trailing comma, otherwise Python treats the parentheses as a standard grouping operator (e.g., an integer or string).
</div>

```python
single_element = (5,)     # Valid tuple
not_a_tuple = (5)         # Evaluates to integer 5
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">3. Without Parentheses (Tuple Packing / Comma Implicit Creation)</span><br>
In Python, it's actually the comma, not parentheses, that defines a tuple.
</div>

```python
tup = 1, 2, 3
single_tup = 4,

print(type(tup))        # Output: <class 'tuple'>
print(type(single_tup)) # Output: <class 'tuple'>
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">4. Using the tuple() Built-in Constructor</span><br>
Converts any iterable (lists, strings, sets, dictionaries, ranges, generators) into a tuple

```python
from_list = tuple([1, 2, 3])
from_string = tuple("Geeks")       # Output: ('G', 'e', 'e', 'k', 's')
from_range = tuple(range(5))       # Output: (0, 1, 2, 3, 4)
from_dict = tuple({'a': 1, 'b': 2}) # Output: ('a', 'b') - converts keys
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">5. Using Unpacking with Asterisk (*)</span><br>
Combines multiple iterables or items into a tuple using comma separated unpacking inside parentheses.
</div>

```python
a = [1, 2]
b = (3, 4)
combined = (*a, *b, 5)  # Output: (1, 2, 3, 4, 5)
print(combined)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">6. Generator Expression Inside tuple()</span><br>
Pass a generator expression to tuple() to generate a tuple dynamically (similar to list comprehension).

```python
squares = tuple(x**2 for x in range(5))  # Output: (0, 1, 4, 9, 16)
print(squares, end=" ")
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">7. Named Tuples (collections.namedtuple)</span><br>
Creates a subclass of tuples where elements can be accessed by attribute name as well as index.

```python
from collections import namedtuple

Point = namedtuple('Point', ['x', 'y', 'z'])
p = Point(10, 20, 30)

print(p.x, p.y, p.z)  # Output: 10 20
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #0e9dc5">Creating a Tuple with Mixed Datatypes</span><br>
Tuples can store elements of different data types, such as integers, strings, lists and dictionaries, within a single structure.

```python
tup = (5, 'Welcome', 7.5, True, [1, 2, 3], {'key': 'value'})
print(tup)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #0e9dc5">Tuple Basic Operations</span><br>
<span style="color: #343F50">1. Accessing Elements (Indexing & Slicing)</span><br>
&emsp; Like lists, tuples support zero-based indexing, negative indexing, and slicing.

```python
tup = (10, 20, 30, 40, 50)

# Single element indexing
print(tup[0])    # Output: 10
print(tup[-1])   # Output: 50 (Last element)

# Slicing: tup[start:stop:step]
print(tup[1:4])  # Output: (20, 30, 40)
print(tup[::-1]) # Output: (50, 40, 30, 20, 10) (Reversed tuple)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">2. Concatenation (+)</span><br>
&emsp; Combines two or more tuples into a new tuple. Note that both operands must be tuples.

```python
tup1 = (1, 2, 3)
tup2 = ('a', 'b')

result = tup1 + tup2
print(result)  # Output: (1, 2, 3, 'a', 'b')
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">3. Repetition (*)</span><br>
&emsp; Duplicates the contents of a tuple a specified number of times.

```python
tup = (1, 2)
result = tup * 3

print(result)  # Output: (1, 2, 1, 2, 1, 2)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">4. Membership Testing (in / not in)</span><br>
&emsp; Checks if an item exists within a tuple.

```python
tup = (10, 20, 30)

print(20 in tup)      # Output: True
print(50 not in tup)  # Output: True
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">5. Tuple Unpacking</span><br>
&emsp; Extracts tuple values directly into separate variables.

```python
# Basic Unpacking
tup = ("Alice", 25, "Developer")
name, age, job = tup

# Extended Unpacking using Asterisk (*)
numbers = (1, 2, 3, 4, 5)
first, *middle, last = numbers
print(first)   # Output: 1
print(middle)  # Output: [2, 3, 4] (Collected into a list)
print(last)    # Output: 5
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">6. Built-In Functions & Methods</span><br>
&emsp; Common functions and methods used with tuples:

```python
tup = (10, 20, 30, 20, 40)

# Built-in Tuple Methods
print(tup.count(20)) # Output: 2 (Occurrences of 20)
print(tup.index(30)) # Output: 2 (First index of 30)

# Built-in Functions
print(len(tup))      # Output: 5 (Length)
print(max(tup))      # Output: 40
print(min(tup))      # Output: 10
print(sum(tup))      # Output: 120
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">7. Deleting a Tuple (del)</span><br>
Because tuples are immutable, individual elements cannot be removed or altered. However, you can delete the entire tuple variable using the del statement.
</div>

```python
tup = (1, 2, 3)
del tup
# print(tup)  # Accessing `tup` after this will raise a NameError
```
