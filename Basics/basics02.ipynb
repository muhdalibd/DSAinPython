<h3 style="color:#fa851e">Python Strings</h3>

- Strings in python are surrounded by either single quotation marks, or double quotation marks.

```python
print("It's alright")
print("He is called 'Johnny'")
print('He is called "Johnny"')
```

<h4 style="color: #008080"> Assign String to a Variable </h4>

- You can assign a multiline string to a variable by using three quotes:

```python
a = "Hello"
a = """Lorem ipsum dolor sit amet, consectetur adipiscing elit"""
a = '''Lorem ipsum dolor sit amet, consectetur adipiscing elit'''
print(a)
```

<h4 style="color: #008080">  Strings are Arrays </h4>

- Python does not have a character data type, a single character is simply a string with a length of 1.

```python
a = "Hello, World!"
print(a[1])
print(len(a))   # String Length
# Looping Through a String
for x in "banana":
    print(x, end=" ")

# Check String: if a certain phrase or character is present in a string
txt = "The best things in life are free!"
print("free" in txt)
if "free" in txt:
    print("Yes, 'free' is present.")

print("expensive" not in txt)
if "expensive" not in txt:
    print("No, 'expensive' is NOT present.")
```

<h4 style="color: #008080"> Python - Slicing Strings </h4>

- You can return a range of characters by using the slice syntax.
- Specify the start index and the end index (not included), separated by a colon, to return a part of the string.

```python
b = "Hello, World!"
print(b[2:5])
print(b[:5])    # Slice From the Start
print(b[2:])    # Slice To the End
print(b[-5:-2])   # Negative Indexing
```

<h4 style="color: #008080"> Python - Modify Strings </h4>

- Python has a set of built-in methods that you can use on strings.

```python
# a = "Hello, World!"
print(a.upper())    # Upper Case
print(a.lower())    # Lower Case
# # The strip() method removes any whitespace from the beginning or the end:
a = " Hello, World! "   # Remove Whitespace
print(a.strip()) # returns "Hello, World!"
a = "Hello, World!"
print(a.replace("H", "J"))  # Replace String

# Split String
# The split() method returns a list.
# The split() method splits the string into substrings if it finds instances of the separator:

a = "Hello, World!"
print(a.split(",")) # returns ['Hello', ' World!']
```

<h4 style="color: #008080"> String Concatenation </h4>

- To concatenate, or combine, two strings you can use the + operator.

```python
a = "Hello"
b = "World"
c = a + b
c = a + " " + b
print(c)
```

<h4 style="color: #008080"> String Format </h4>

- We cannot combine strings and numbers like this:
- F-String was introduced in Python 3.6, and is now the preferred way of formatting strings.
- To specify a string as an f-string, simply put an f in front of the string literal, and add curly brackets { } as placeholders for variables and other operations.

```python
age = 36
# txt = "My name is John, I am " + age    # TypeError: can only concatenate str (not "int") to str
txt = f"My name is John, I am {age}"
print(txt)
```

<h4 style="color: #008080"> Placeholders and Modifiers </h4>

- A placeholder can contain variables, operations, functions, and modifiers to format the value.

```python
price = 59
txt = f"The price is {price:.2f} dollars"
print(txt)
txt = f"The price is {20 * 59} dollars"     # math operation
print(txt)
```

<h4 style="color: #008080"> Escape Character </h4>

- An escape character is a backslash \ followed by the character you want to insert.
<pre>
   Code       Result

    \'	    Single Quote
    \\	    Backslash	 
    \n	    New Line	
    \r	    Carriage Return	
    \t	    Tab	
    \b	    Backspace	
    \f	    Form Feed	
    \ooo    Octal value	
    \xhh    Hex value
</pre>

```python
txt = "We are the so-called \"Vikings\" from the north."
print(txt)
txt = 'We are the so-called "Vikings" from the north.'
print(txt)
```
