<h3 style="color:#fa851e"> Python Booleans </h3>

- Booleans represent one of two values: True or False.

```python
print(bool(False))
print(bool(None))
print(bool(0))
print(bool(""))
print(bool(()))
print(bool([]))
print(bool({}))
```

<h4 style="color: #008080"> Assignment Operators </h4>
<pre>
Operator    Example	Same As
    =	    x = 5	x = 5	
    +=      x += 3	x = x + 3	
    -=	    x -= 3	x = x - 3	
    *=	    x *= 3	x = x * 3	
    /=	    x /= 3	x = x / 3	
    %=	    x %= 3	x = x % 3	
    //=     x //= 3	x = x // 3	
    **=	    x **= 3	x = x ** 3	
    &=	    x &= 3	x = x & 3	
    |=	    x |= 3	x = x | 3	
    ^=	    x ^= 3	x = x ^ 3	
    >>=	    x >>= 3	x = x >> 3	
    <<=	    x <<= 3	x = x << 3	
    :=	   print(x:=3)	 x = 3
                        print(x)
</pre>


```python
# Python 3.8 introduced the := operator, known as the "walrus operator".
# It assigns values to variables as part of a larger expression:
numbers = [1, 2, 3, 4, 5]
if (count := len(numbers)) > 3:
    print(f"List has {count} elements")
```

<h4 style="color: #008080"> Ternary Operator </h4>

- The ternary operator allows you to assign one value if a condition is true, and another if it is false:

```python
num = 6
x = "WEEKEND!" if num > 5 else "Workday"
print(x)
num = 6
x = "Fri" if num == 5 else "Sat" if num == 6 else "Sun" if num == 7 else "weekday"
print(x)
```

<h4 style="color:#008080"> Identity Operators </h4>

- Identity operators are used to compare the objects, not if they are equal, but if they are actually the same object, with the same memory location:
<pre>
  Operator	Description                                             Example
    is          Returns True if both variables are the same object	x is y	
    is not	Returns True if both variables are not the same object	x is not y
</pre>
<h4 style="color: #008080"> Difference Between is and == </h4>

- is  -  Checks if both variables point to the same object in memory <br>
- ==  -  Checks if the values of both variables are equal
