<h4 style="color:#fa851e">Python Lists</h4>
<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
List is a built-in data structure used to store an ordered collection of items. They are dynamic, resizable and capable of storing multiple data types.
<pre style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
    Mutable: list elements can be changed, updated, added, or removed after the list is created.
    Ordered: elements maintain the order in which they are inserted.
    Index-based: elements are accessed using their position, starting from index 0.
</pre>
</div>

<h5 style="color:#0e9dc5">Creating a List</h5>
<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
1. Using Square Brackets [] (Literal Syntax):<br>
The most common, direct, and concise way to define a list.
</div>

```python
empty_list = []
mixed_list = [1, "apple", 3.14, True]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
2. Using the list() Constructor:<br>
Converts any iterable (tuples, strings, sets, dictionaries, ranges, generators, etc.) into a list.
</div>

```python
empty_list = list()
from_tuple = list((1, 2, 3))
from_string = list("hello")  # Output: ['h', 'e', 'l', 'l', 'o']
from_range = list(range(5))   # Output: [0, 1, 2, 3, 4]
from_dict = list({'a': 1, 'b': 2})  # Converts keys -> ['a', 'b']
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
3. List Comprehension:<br>
Creates a list dynamically by applying an expression to each item in an iterable, optionally filtering with conditions.
</div>

```python
squares = [x**2 for x in range(5)]  # Output: [0, 1, 4, 9, 16]
evens = [x for x in range(10) if x % 2 == 0]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
4. Repetition Operator (*):<br>
Repeats elements a specified number of times to pre-fill a list.
</div>

```python
zeros = [0] * 5  # Output: [0, 0, 0, 0, 0]
pattern = ["a", "b"] * 3  # Output: ['a', 'b', 'a', 'b', 'a', 'b']
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
5. Slicing an Existing List:<br>
Creates a shallow copy or sub-list from an existing list.
</div>

```python
original = [1, 2, 3, 4, 5]
copy_list = original[:]
sub_list = original[1:4]  # Output: [2, 3, 4]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
6. Unpacking Operator (*):<br>
Uses the iterable unpacking operator inside square brackets to create or combine lists.
</div>

```python
a = [1, 2]
b = [3, 4]
combined = [*a, *b]  # Output: [1, 2, 3, 4]
from_tuple = [*("x", "y")]  # Output: ['x', 'y']
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
7. Standard Library Modules:<br>
copy.copy() or copy.deepcopy(): Creates a copy of an existing list.
</div>

```python
import copy
new_list = copy.copy([1, 2, 3])
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
itertools module: Generates lists from combinations, permutations, or infinite sequences.
</div>

```python
import itertools
combos = list(itertools.combinations([1, 2, 3], 2))  # Output: [(1, 2), (1, 3), (2, 3)]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
array or collections converting to list:
</div>

```python
from collections import deque
d = deque([1, 2, 3])
lst = list(d)
```

<h5 style="color:#0e9dc5">Internal Representation of Lists</h5>
<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
Python list stores references to objects, not the actual values directly.
<pre style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
    The list keeps memory addresses of objects like integers, strings or booleans.
    Actual objects exist separately in memory.
    Modifying a mutable object inside a list changes the original object.
    Reassigning an immutable object creates a new object instead of changing the old one.
</pre>
</div>
