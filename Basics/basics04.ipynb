<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color:#fa851e; font-size: 20px">Python Lists</span><br>
List is a built-in data structure used to store an ordered collection of items. They are dynamic, resizable and capable of storing multiple data types.<br>
&emsp; Mutable: list elements can be changed, updated, added, or removed after the list is created.<br>
&emsp; Ordered: elements maintain the order in which they are inserted.<br>
&emsp; Index-based: elements are accessed using their position, starting from index 0.
</div>

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #0e9dc5">Creating a List</span><br>
<span style="color: #343F50">1. Using Square Brackets [] (Literal Syntax):</span><br>
&emsp; The most common, direct, and concise way to define a list.
</div>

```python
empty_list = []
mixed_list = [1, "apple", 3.14, True]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">2. Using the list() Constructor:</span><br>
&emsp; Converts any iterable (tuples, strings, sets, dictionaries, ranges, generators, etc.) into a list.
</div>

```python
empty_list = list()
from_tuple = list((1, 2, 3))
from_string = list("hello")  # Output: ['h', 'e', 'l', 'l', 'o']
from_range = list(range(5))   # Output: [0, 1, 2, 3, 4]
from_dict = list({'a': 1, 'b': 2})  # Converts keys -> ['a', 'b']
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">3. List Comprehension:</span><br>
&emsp; Creates a list dynamically by applying an expression to each item in an iterable, optionally filtering with conditions.
</div>

```python
squares = [x**2 for x in range(5)]  # Output: [0, 1, 4, 9, 16]
evens = [x for x in range(10) if x % 2 == 0]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">4. Repetition Operator (*):</span><br>
&emsp; Repeats elements a specified number of times to pre-fill a list.
</div>

```python
zeros = [0] * 5  # Output: [0, 0, 0, 0, 0]
pattern = ["a", "b"] * 3  # Output: ['a', 'b', 'a', 'b', 'a', 'b']
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">5. Slicing an Existing List:</span><br>
&emsp; Creates a shallow copy or sub-list from an existing list.
</div>

```python
original = [1, 2, 3, 4, 5]
copy_list = original[:]
sub_list = original[1:4]  # Output: [2, 3, 4]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">6. Unpacking Operator (*):</span><br>
&emsp; Uses the iterable unpacking operator inside square brackets to create or combine lists.
</div>

```python
a = [1, 2]
b = [3, 4]
combined = [*a, *b]  # Output: [1, 2, 3, 4]
from_tuple = [*("x", "y")]  # Output: ['x', 'y']
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color:#0e9dc5">Internal Representation of Lists</span><br>
Python list stores references to objects, not the actual values directly.<br>
&emsp; The list keeps memory addresses of objects like integers, strings or booleans.<br>
&emsp; Actual objects exist separately in memory.<br>
&emsp; Modifying a mutable object inside a list changes the original object.<br>
&emsp; Reassigning an immutable object creates a new object instead of changing the old one.
</div>

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #0e9dc5">Accessing Elements</span><br>
Elements in a list are accessed using indexing. Python uses zero-based indexing, meaning a[0] represents the first element. Negative indexing is also supported, where -1 accesses the last element.<br>
<span style="color: #343F50">1. Basic Indexing (Zero-Based)</span><br>
&emsp; The first element starts at index 0.
</div>

```python
a = [10, 20, 30, 40]
print(a[0])  # Output: 10 (First element)
print(a[2])  # Output: 30 (Third element)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">2. Negative Indexing</span><br>
&emsp; Negative indices count backward from the end of the list (-1 refers to the last element).

```python
a = [10, 20, 30, 40]
print(a[-1])  # Output: 40 (Last element)
print(a[-2])  # Output: 30 (Second to last element)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">3. List Slicing ([start:stop:step])</span><br>
&emsp; To extract a sub-list, use slicing. The stop index is non-inclusive.

```python
a = [10, 20, 30, 40, 50, 60]

print(a[1:4])   # Output: [20, 30, 40] (Indices 1 to 3)
print(a[:3])    # Output: [10, 20, 30] (From start to index 2)
print(a[3:])    # Output: [40, 50, 60] (From index 3 to end)
print(a[::2])   # Output: [10, 30, 50] (Every 2nd element)
print(a[::-1])  # Output: [60, 50, 40, 30, 20, 10] (Reversed list)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">4. Accessing Elements in Nested Lists</span><br>
&emsp; Chained indexing is used to access elements inside multidimensional lists.

```python
matrix = [[1, 2], [3, 4]]

print(matrix[0])     # Output: [1, 2] (First sub-list)
print(matrix[1][0])  # Output: 3 (Second sub-list, first element)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #0e9dc5">Adding Elements</span><br>
&emsp; Elements can be added to a list using the following methods:<br>
<span style="color: #343F50">1. append()</span><br>
&emsp; Adds a single element to the very end of the list.<br>
&emsp; Note: If you pass a list to append(), it adds the entire list as a single nested element (e.g., [1, 2, [3, 4]]).
</div>

```python
a = [1, 2]
a.append(3)
a.append([4, 5, 6])

print(a)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">2. insert()</span><br>
&emsp; Adds an element at a specific index position. Shifts existing elements to the right.

```python
a = [1, 3]
a.insert(1, 2)  # Syntax: insert(index, element)

print(a)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">3. extend()</span><br>
&emsp; Iterates over its argument and adds each item individually to the end of the list.<br>
&emsp; Useful for merging iterables (lists, tuples, sets) into an existing list.

```python
a = [1, 2]
a.extend([3, 4])    # Extending with a List
a.extend((5, 6))    # Extending with a Tuple
# a.extend((8))     # TypeError: 'int' object is not iterable
a.extend({70, 8})   # Extending with a Set
# Unpredictable Order: The exact order of 70 and 8 at the end of the list 
# is not guaranteed because sets do not maintain insertion order.
print(a)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #0e9dc5">1. Updating Elements</span><br>
Since Python lists are mutable, you can modify existing elements by referencing their indices or slices.<br>
<span style="color: #343F50">A. Updating a Single Element (Indexing)</span><br>
&emsp; Assign a new value to a specific index.
</div>

```python
a = [10, 20, 30, 40, 50]
a[1] = 25

print(a)  # Output: [10, 25, 30, 40, 50]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">B. Updating Multiple Elements (Slicing)</span><br>
&emsp; Replace a sub-range of elements with a new iterable.

```python
a = [10, 20, 30, 40, 50]
a[1:4] = [200, 300]  # Replaces elements at index 1, 2, and 3

print(a)  # Output: [10, 200, 300, 50]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #0e9dc5">2. Removing Elements</span><br>
Python provides several ways to delete items depending on whether you know the value, the index position, or want to clear the entire list.<br>
<span style="color: #343F50">A. remove() — By Value</span><br>
&emsp; Removes the first occurrence of a specified value. Raises a ValueError if the value is not found.

```python
a = [1, 2, 3, 2]
a.remove(2)

print(a)  # Output: [1, 3, 2]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">B. pop() — By Index (Returns Removed Value)</span><br>
&emsp; Removes and returns the element at the specified index.<br>
&emsp; If no index is provided, it removes and returns the last element.

```python
a = [10, 20, 30, 40]

last_item = a.pop()      # Removes last element (40)
second_item = a.pop(1)   # Removes element at index 1 (20)

print(a)           # Output: [10, 30]
print(last_item)   # Output: 40
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">C. del Statement — By Index or Slice</span><br>
&emsp; Deletes an element at a specific index or deletes a slice of elements without returning the values.

```python
a = [10, 20, 30, 40, 50]

del a[1]       # Removes element at index 1
print(a)       # Output: [10, 30, 40, 50]

del a[1:3]     # Removes elements at indices 1 and 2
print(a)       # Output: [10, 50]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">D. clear() — Remove All Elements</span><br>
&emsp; Empties the entire list, leaving it as an empty list [].

```python
a = [1, 2, 3]
a.clear()

print(a)  # Output: []
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #0e9dc5">Iterating Over Lists</span><br>
Lists can be iterated using loops, allowing operations to be performed on each element.<br>
<span style="color: #343F50">1. Simple for Loop (Direct Iteration)</span><br>
&emsp; Iterates directly through the elements of the list.

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">2. Using range() and len() (Index-Based Iteration)</span><br>
&emsp; Use this when you need to access or modify items using their positional index.

```python
fruits = ["apple", "banana", "cherry"]
for i in range(len(fruits)):
    print(i, fruits[i])
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">3. Using enumerate() (Index and Element Pair)</span><br>
&emsp; The pythonic way to access both the index and the value at the same time.

```python
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(f"Index {index}: {fruit}")
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">4. While Loop</span><br>
&emsp; Iterates manually using a counter index variable.

```python
fruits = ["apple", "banana", "cherry"]
i = 0
while i < len(fruits):
    print(fruits[i])
    i += 1
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">5. List Comprehension (Inline Iteration)</span><br>
&emsp; Useful for performing an action or transformation on every item to generate a new list.

```python
fruits = ["apple", "banana", "cherry"]

# Print squared lengths or uppercase strings inline
uppercase_fruits = [fruit.upper() for fruit in fruits]
print(uppercase_fruits)  # Output: ['APPLE', 'BANANA', 'CHERRY']
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">6. Using zip() (Iterating Multiple Lists Simultaneously)</span><br>
&emsp; Iterates over two or more lists at the same time in parallel.

```python
names = ["Alice", "Bob"]
scores = [85, 92]

for name, score in zip(names, scores):
    print(f"{name} scored {score}")
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #0e9dc5">Nested Lists</span><br>
A nested list is a list that contains another list (or multiple lists) as its elements. They are commonly used to represent multi-dimensional data, matrices, grids, or hierarchical structures.<br>
<span style="color: #343F50">1. Creating a Nested List</span><br>
&emsp; You define a nested list by placing standard list brackets inside an outer list.
</div>

```python
# A 2D matrix (2 rows, 3 columns)
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">2. Accessing Elements</span><br>
Accessing elements in a nested list requires chained indexing ([row][column]):<br>
&emsp; The first index selects the inner list.<br>
&emsp; The second index selects the item inside that inner list.<br>
</div>

```python
matrix = [
    [10, 20, 30],
    [40, 50, 60]
]

# Access the first row (returns a list)
print(matrix[0])  # Output: [10, 20, 30]

# Access element at row 1, column 2 (0-indexed)
print(matrix[1][2])  # Output: 60
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">3. Iterating Over Nested Lists</span><br>
&emsp; Use nested loops to traverse through all sub-elements.
</div>

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

for row in matrix:
    for item in row:
        print(item, end=" ")
    print()  # Newline after each row
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">4. Modifying Nested Lists</span><br>
&emsp; Assign a new value to a specific inner index position.
</div>

```python
matrix = [
    [1, 2],
    [3, 4]
]
matrix[0][1] = 99
print(matrix)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">5. Flattening a Nested List (List Comprehension)</span><br>
&emsp; Flatten a 2D list into a single 1D list using double list comprehension.
</div>

```python
matrix = [[1, 2], [3, 4], [5, 6]]
flattened = [item for row in matrix for item in row]
print(flattened)  # Output: [1, 2, 3, 4, 5, 6]
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">⚠️ Common Pitfall: List Multiplication with Sub-lists</span><br>
When initializing a nested list using multiplication [ [] ] * n, Python creates multiple references to the same inner list.
</div>

```python
# WRONG WAY:
grid = [[0] * 3] * 3
grid[0][0] = 1
print(grid)  # Output: [[1, 0, 0], [1, 0, 0], [1, 0, 0]] (Modifies all rows!)

# RIGHT WAY (List Comprehension):
grid = [[0] * 3 for _ in range(3)]
grid[0][0] = 1
print(grid)  # Output: [[1, 0, 0], [0, 0, 0], [0, 0, 0]]
```
