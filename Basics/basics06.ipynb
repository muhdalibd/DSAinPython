<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color:#fa851e; font-size: 20px">Python Dictionary</span><br>
Dictionary is a data structure that stores information in key-value pairs. While keys must be unique and immutable (like strings or numbers), values can be of any data type, whether mutable or immutable. This makes dictionaries ideal for accessing data by a specific name rather than a numeric position like in list.
</div>

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #0e9dc5">Creating a Dictionary</span><br>
A dictionary is created by writing key-value pairs inside { }, where each key is connected to a value using colon (:).<br>
<span style="color: #343F50">1. Using Curly Braces {} (Literal Syntax)</span><br>
The most common, direct, and readable way to create a dictionary using key-value pairs.
</div>

```python
empty_dict = {}
person = {"name": "Alice", "age": 25, "city": "New York"}
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">2. Using the dict() Constructor</span><br>
The built-in constructor offers multiple ways to initialize a dictionary:<br>
<span style="color: #343F50">A. Keyword Arguments</span><br>
Keys are passed as named parameters (keys must be valid Python identifiers/strings).

```python
empty_dict = dict()
person = dict(name="Alice", age=25, city="New York")
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">B. List of Tuples (Key-Value Pairs)</span><br>
&emsp; Pass a list or sequence of (key, value) tuples.

```python
pairs = [("name", "Alice"), ("age", 25)]
person = dict(pairs)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">3. Dictionary Comprehension</span><br>
Dynamically construct a dictionary from an iterable applying transformations or conditional filtering.
</div>

```python
# Square of numbers from 0 to 4
squares = {x: x**2 for x in range(5)}
# Output: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

# Filtering even numbers
evens_map = {x: x for x in range(10) if x % 2 == 0}
print(evens_map)
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">4. Using dict.fromkeys()</span><br>
Creates a new dictionary with specified keys and a uniform default value.
</div>

```python
keys = ["a", "b", "c"]

# With default value None
d1 = dict.fromkeys(keys)  # Output: {'a': None, 'b': None, 'c': None}

# With a custom initial value
d2 = dict.fromkeys(keys, 0)  # Output: {'a': 0, 'b': 0, 'c': 0}
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">5. Using zip() with dict()</span><br>
Combines two separate iterables (one for keys and one for values) into key-value pairs.
</div>

```python
keys = ["name", "age", "role"]
values = ["Bob", 30, "Developer"]

person = dict(zip(keys, values))
# Output: {'name': 'Bob', 'age': 30, 'role': 'Developer'}
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">6. Unpacking Operator (**)</span><br>
Merges existing dictionaries or key-value structures into a new dictionary using dictionary unpacking.
</div>

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

merged = {**dict1, **dict2}
# Output: {'a': 1, 'b': 2, 'c': 3, 'd': 4}
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">7. Dictionary Union Operator (|)</span><br>
Available in Python 3.9+, this operator merges two dictionaries to produce a new dictionary.
</div>

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 99, "c": 3}  # 'b' in dict2 will overwrite 'b' in dict1

merged = dict1 | dict2
# Output: {'a': 1, 'b': 99, 'c': 3}
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">8. Special Dictionary Types (collections module)</span><br>
Standard library modules offer dictionary variants for specific use cases:<br>
<span style="color: #343F50">A. defaultdict</span><br>
Automatically provides a default value when accessing a non-existent key.
</div>

```python
from collections import defaultdict

d = defaultdict(list)  # Missing keys auto-initialize with an empty list []
d["items"].append("apple")
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">B. OrderedDict</span><br>
Remembers the insertion order of keys (note: standard dict in Python 3.7+ also preserves insertion order, but OrderedDict provides specialized methods like move_to_end()).
</div>

```python
from collections import OrderedDict

d = OrderedDict([("a", 1), ("b", 2)])
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #0e9dc5">Accessing Dictionary Items</span><br>
A value in a dictionary is accessed by using its key.
</div>

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">1. Using Square Bracket Notation ([])</span><br>
Access a value directly using its key inside square brackets.<br>
⚠️ Note: If the specified key does not exist, this method raises a KeyError.

```python
person = {"name": "Kat", "age": 21}

print(person["name"])  # Output: Kat
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">2. Using the .get() Method (Safe Access)</span><br>
Returns the value for the given key if present. If the key is not found, it safely returns None (or a custom default value) without raising an error.
</div>

```python
person = {"name": "Kat", "age": 21}

# Basic usage
print(person.get("age"))      # Output: 21
print(person.get("salary"))   # Output: None

# Providing a custom default value
print(person.get("salary", 0)) # Output: 0
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">3. Accessing All Keys (.keys())</span><br>
Returns a dynamic view object containing all the keys in the dictionary.
</div>

```python
person = {"name": "Kat", "age": 21}

print(person.keys())  # Output: dict_keys(['name', 'age'])

# Convert to a list if needed
keys_list = list(person.keys())  # ['name', 'age']
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">4. Accessing All Values (.values())</span><br>
Returns a view object containing all the values in the dictionary.
</div>

```python
person = {"name": "Kat", "age": 21}

print(person.values())  # Output: dict_values(['Kat', 21])
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">5. Accessing Key-Value Pairs (.items())</span><br>
Returns a view object containing each (key, value) pair as a tuple. Excellent for iterating in for loops.
</div>

```python
person = {"name": "Kat", "age": 21}

print(person.items())  # Output: dict_items([('name', 'Kat'), ('age', 21)])

# Unpacking in a loop
for key, value in person.items():
    print(f"{key}: {value}")
```

<div style="font-family: Comic Sans MS; font-weight: 400; font-size: 16px; color: gray;">
<span style="color: #343F50">6. Accessing Elements in Nested Dictionaries</span><br>
Use chained square brackets or chained .get() calls to traverse multi-level dictionaries.
</div>

```python
data = {
    "student": {
        "name": "Sam",
        "scores": {"math": 95}
    }
}

# Chained brackets
print(data["student"]["name"])           # Output: Sam
print(data["student"]["scores"]["math"]) # Output: 95

# Safe chained access with .get()
print(data.get("student", {}).get("name"))  # Output: Sam
```
