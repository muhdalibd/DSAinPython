<h3 style="color:#fa851e">Python Statements</h3>

- A computer program is a list of "instructions" to be "executed" by a computer.
- In a programming language, these programming instructions are called statements.

```python
print("Hello World!")
print("Have a good day.")
print("Learning Python is fun!")
```

<h4 style="color: #008080">Semicolons are optional in Python. </h4>

- You can write multiple statements on one line by separating them with ; but this is rarely used because it makes it hard to read:

```python
print("Hello"); print("How are you?"); print("Bye bye!")
# print("Python is fun!") print("Really!")    # SyntaxError: invalid syntax
```

<h3 style="color:#fa851e">Python Output / Print</h3>

- Text in Python must be inside quotes.
- You can use either " double quotes or ' single quotes (1 or 3 times):

```python
print("This will work!")
print('This will also work!')
print('''This will also work!''')
print('Bangladesh', 2.0, "36 July")
# print(This will cause an error)     # SyntaxError: invalid syntax.
```

<h4 style="color: #008080"> Print Without a New Line <h4>

##### By default, the print() function ends with a new line. If you want to print multiple words on the same line, you can use the end parameter:

```python
print("Hello World!", end=" ")  # space after end=" " for better readability.
print("I will print on the same line.")
print("Learn Basics", end=" --> ")
print("Learn DSA")
```

<h3 style="color:#fa851e">Python Variables</h3>

- Variables are containers for storing data values.
- A variable is created the moment you first assign a value to it.
- Variables do not need to be declared with any particular type, and can even change type after they have been set.

```python
x = 5       # x is of type int
y = "John"  # y is of type str
print(x)
print(y)
x = "Sally" # x is now of type str
print(x)
```

<h4 style="color: #008080"> Casting </h4>

- If you want to specify the data type of a variable, this can be done with casting.

```python
x = str(3)    # x will be '3'
y = int(3)    # y will be 3
z = float(3)  # z will be 3.0
```

<h4 style="color: #008080">Get the Type </h4>

- You can get the data type of a variable with the type() function.

```python
x = 5
y = "John"
print(type(x))
print(type(y))
```

<h4 style="color: #008080">Variable Names </h4>

A variable can have a short name (like x and y) or a more descriptive name (age, carname, total_volume).
Rules for Python variables:
- A variable name can only contain letter, number and underscores (A-z, 0-9, and _ )
- A variable name must start with a letter or the underscore character
- A variable name cannot start with a number
- Variable names are <span style="color: #008080"> case-sensitive </span> (age, Age and AGE are three different variables)
- A variable name cannot be any of the Python keywords.

```python
# Illegal variable names:
# 2myvar = "John"
# my-var = "John"
# my var = "John"
```

<h4 style="color: #008080">Multi Words Variable Names </h4>

Variable names must be meaningful. There are several techniques make them more readable:
- Camel Case :<br>
&emsp;    myVariableName = "John"
- Pascal Case :<br>
&emsp;    MyVariableName = "John"
- Snake Case :<br>
&emsp;    my_variable_name = "John"

<h4 style="color: #008080">Many Values to Multiple Variables </h4>

- Python allows you to assign values to multiple variables in one line:


```python
x, y, z = "Orange", "Banana", "Cherry"
print(x)
print(y)
print(z)
```

<h4 style="color: #008080">One Value to Multiple Variables </h4>

- And you can assign the same value to multiple variables in one line:

```python
x = y = z = "Orange"
print(x)
print(y)
print(z)
```

<h4 style="color: #008080">Unpack a Collection </h4>

- If you have a collection of values in a list, tuple etc.
- Python allows you to extract the values into variables. This is called unpacking.

```python
fruits = ["apple", "banana", "cherry"]
x, y, z = fruits
print(x)
print(y)
print(z)
```

<h4 style="color: #008080"> Output Variables </h4>

```python
x = "Python"
y = "is"
z = "awesome"
print(x, y, z)
print(x + y + z)
print(x + " " + y + " " + z)
```

<h4 style="color: #008080">Global Variables </h4>

- Variables that are created outside of a function (as in all of the examples in the previous shells) are known as global variables.
- Global variables can be used by everyone, both inside of functions and outside.

<h3 style="color: #fa851e">Python Data Types </h3>
Built-in Data Types <br>
Python has the following data types built-in by default, in these categories:

- Text Type:&emsp;&emsp;    str
- Numeric Types:&emsp;  int, float, complex
- Sequence Types:&emsp; list, tuple, range
- Mapping Type:&emsp;   dict
- Set Types:&emsp; &emsp;   set, frozenset
- Boolean Type:&emsp;   bool
- Binary Types:&emsp;   bytes, bytearray, memoryview
- None Type:&emsp;  NoneType

```python
# Float can also be scientific numbers with an "e" to indicate the power of 10.
x = 35e3
y = 12E4
z = -87.7e100

print(type(x))
print(type(y))
print(type(z))
```
